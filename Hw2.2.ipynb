```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('2017_jun_final.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.dtypes
```

```python
df.isnull().sum() / df.shape[0]
```

```python
null_share = df.isnull().sum() / df.shape[0]
cols = null_share[null_share > 0].index.drop('Мова програмування')
df = df.drop(cols, axis=1)
```

```python
df.isnull().sum() / df.shape[0]
```

```python
df = df.dropna()
```

```python
df.shape
```

```python
python_data = df[df['Мова програмування'] == 'Python']
```

```python
python_data.shape
```

```python
grouped = python_data.groupby('Посада')
grouped.size()
```

```python
salary = grouped['Зарплата на місяць'].agg(['min', 'max'])
salary
```

```python
def fill_avg_salary(row):
    return (row['min'] + row['max']) / 2

salary['avg'] = salary.apply(fill_avg_salary, axis=1)
salary
```

```python
salary['avg'].describe()
```

```python
salary.to_csv('python_salary.csv')
```

```python
top = df['Мова програмування'].value_counts().head(10)
sns.set_style('darkgrid')
sns.barplot(x=top.values, y=top.index, hue=top.index, palette='viridis', legend=False)
plt.title('топ-10 мов програмування')
plt.show()
```

```python
positions = python_data['Посада'].value_counts()
plt.pie(positions, autopct=lambda p: f'{p:.1f}%' if p > 3 else '', colors=sns.color_palette('pastel'), shadow=True)
plt.legend(positions.index, loc='center left', bbox_to_anchor=(1, 0.5))
plt.title('посади python-розробників')
plt.show()
```

```python
sns.set_style('whitegrid')
sns.boxplot(x='Зарплата на місяць', y='Посада', data=python_data, hue='Посада', palette='Set2', legend=False)
plt.title('зарплата python-розробників за посадами')
plt.show()
```

```python
plt.hist(python_data['Зарплата на місяць'], bins=25, color='#FF8C42', edgecolor='white')
plt.axvline(python_data['Зарплата на місяць'].mean(), color='navy', linestyle='--', label='середня')
plt.axvline(python_data['Зарплата на місяць'].median(), color='green', linestyle=':', label='медіана')
plt.legend()
plt.title('розподіл зарплат python-розробників')
plt.show()
```

```python
sns.set_style('white')
sns.regplot(x='exp', y='Зарплата на місяць', data=python_data, scatter_kws={'alpha': 0.5, 'color': 'purple'}, line_kws={'color': 'red'})
plt.title('досвід vs зарплата')
plt.show()
```
